### 2. Categories of Data Science Tasks & Tools
To efficiently transform raw data into insights, tasks are divided into several distinct phases supported by open-source and commercial tooling:

*   **Data Management:** Storing, organizing, and retrieving structured/unstructured data.
*   **Data Integration and Transformation (ETL):** Extracting data from multiple sources, transforming it, and loading it into central repositories.
*   **Data Visualization:** Representing data graphically via charts and maps to facilitate quick decision-making.
*   **Model Building & Deployment:** Training machine learning algorithms and integrating them into production environments via APIs.
*   **Model Monitoring & Assessment:** Continuously evaluating active models for accuracy, drift, and fairness.
*   **Asset Management (Code & Data):** Collaborative version control for codebases and metadata governance for data cataloging.

## 💻 Module 2: Languages of Data Science

### 1. Primary Languages for Data Scientists
*   **Python:** The absolute leader (used in >75% of jobs). Renowned for its simple, readable syntax and massive package ecosystem (Pandas, NumPy, Scikit-Learn).
*   **R Language:** A free, array-oriented software environment favored by statisticians. Highly effective for complex mathematical translations, data visualization (`ggplot2`), and academic research.
*   **SQL (Structured Query Language):** A non-procedural language designed specifically to query and manage relational database systems. It is the crucial bridge between code and enterprise storage.

### 2. Specialized Languages & Libraries
*   **Scala & Java:** High-performance, object-oriented languages. Java provides tools like Weka and Deeplearning4, while Scala powers **Apache Spark** (essential for high-volume stream/batch processing).
*   **C++:** Offers ultimate execution speed and control; forms the backend of major frameworks like TensorFlow, though users usually interact with it via Python APIs.
*   **Julia:** Engineered for high-performance numerical computing and fast execution. Integrates seamlessly with other languages and uses `JuliaDB` for massive datasets.
*   **JavaScript:** Primarily web-focused, but increasingly popular in ML via `TensorFlow.js` for running browser-based model inferences.

## 📚 Module 3: Libraries, APIs, Datasets, & ML Models

### 1. Python Libraries Ecosystem
*   **Scientific Computing:** `Pandas` (DataFrames for cleaning & manipulating structured data) & `NumPy` (efficient multidimensional array and matrix computations).
*   **Data Visualization:** `Matplotlib` (highly customizable static plotting) & `Seaborn` (modern, out-of-the-box statistical visualizations like heatmaps and violin plots).
*   **Machine & Deep Learning:** `Scikit-learn` (statistical modeling, clustering, and regression) alongside production-grade neural network libraries like `Keras`, `TensorFlow`, and `PyTorch`.

### 2. APIs (Application Programming Interfaces)
An API acts as an intermediary, enabling different software systems to interact without exposing internal source code.
*   **Library APIs:** E.g., Using Python code to access high-performance C++ backend operations inside TensorFlow.
*   **REST APIs:** Communication over HTTP endpoints using standardized requests/responses (usually formatted in JSON). Examples include **Watson Speech-to-Text** and **Watson Language Translator**.

### 3. Data Licensing Reference Guide
When choosing a dataset for your projects, understanding its license terms is essential:

1.  **Public Domain (PDM / PDDL):** Free for anyone to use, modify, and distribute without restrictions.
2.  **Attribution (CC-BY / ODC-BY):** Free to share and adapt, but credit must be given to the creator.
3.  **ShareAlike (CC-BY-SA / ODC-ODbL):** Derived works must be shared under the exact same license.
4.  **Copyleft (CDLA-Sharing-1.0):** Allows modification and distribution, provided you do not restrict downstream users.
5.  **Permissive (CDLA-Permissive-2.0):** Allows wide modification and use as long as a disclaimer of liability is kept.
6.  **Restrictive (CC BY-NC / CC BY-ND):** Restricts use for commercial purposes, or forbids modifying the original dataset.
